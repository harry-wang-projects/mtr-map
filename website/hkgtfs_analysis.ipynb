```python
def busroute_lookup(bus_number, company):
    specific_routes = routes[(routes["route_short_name"] == bus_number) & (routes["agency_id"] == company)]
    for row in specific_routes.itertuples():
        specific_trips = pd.merge(trips[trips["route_id"] == row.route_id], frequencies, on='trip_id', how="left")
        #specific_trips = trips[trips["route_id"] == row.route_id]
        print(specific_trips)
        filtered_stoptimes = stop_times[stop_times["trip_id"] == specific_trips["trip_id"].iloc[0]]
        filtered_stops = filtered_stoptimes.merge(stops[['stop_id', 'stop_name', 'stop_lat', 'stop_lon']], on='stop_id', how='left')
        print(filtered_stops[["stop_name"]])
        print(specific_trips.service_id.unique())

```

```python
busroute_lookup("54", "GMB")
```
